**1.Import**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture

sns.set_theme(style="whitegrid")
RANDOM_STATE = 42
```

**1b.Load the three datasets**

```python
# Wine comes built into scikit-learn
wine_raw = load_wine(as_frame=True)
wine = wine_raw.frame.copy()
wine["target"] = wine["target"].map(dict(enumerate(wine_raw.target_names)))

# Chickwts and USArrests are downloaded from the internet
base = "https://vincentarelbundock.github.io/Rdatasets/csv/datasets/"
chick = pd.read_csv(base + "chickwts.csv", index_col=0)
arrests = pd.read_csv(base + "USArrests.csv", index_col=0)

print("Wine:", wine.shape)
print("Chickwts:", chick.shape)
print("USArrests:", arrests.shape)
```

**1c: Inspect the data**

```python
datasets = {"Wine": wine, "Chickwts": chick, "USArrests": arrests}

for name, df in datasets.items():
    print(f"===== {name} =====")
    print("Shape:", df.shape)
    print("Missing values per column:")
    print(df.isna().sum().to_string())
    print("Duplicate rows:", df.duplicated().sum())
    print("Column types:")
    print(df.dtypes.to_string())
    display(df.head())
    print()
```

**1d: Look at the duplicate before deciding**

```python
# Show every row involved in the duplicate
print(chick[chick.duplicated(keep=False)].sort_values(["feed", "weight"]))

# Count chicks per feed type, and check the feed names are consistent
print(chick["feed"].value_counts())
print(chick["feed"].unique())
```

Chickwts has 1 duplicate row (chicks 25 and 36, both soybean, 248 g). They are different chicks with the same weight, so they were kept. No missing values were found in any dataset, and feed names are consistent.

**1e: Standardize the numbers**

```python
# Wine: scale the 13 measurements (the class name is text, so leave it out)
wine_features = wine.drop(columns="target")
wine_scaled = pd.DataFrame(
    StandardScaler().fit_transform(wine_features),
    columns=wine_features.columns
)

# USArrests: scale all 4 columns, keep the state names as the index
arrests_scaled = pd.DataFrame(
    StandardScaler().fit_transform(arrests),
    columns=arrests.columns,
    index=arrests.index
)

# Chickwts: scale only the weight column
chick["weight_scaled"] = StandardScaler().fit_transform(chick[["weight"]])

# Check: means should be about 0 and standard deviations about 1
print(wine_scaled.describe().loc[["mean", "std"]].round(2))
print(arrests_scaled.describe().loc[["mean", "std"]].round(2))
print(chick["weight_scaled"].agg(["mean", "std"]).round(2))
```

The scaling worked. Every mean is about 0 (the -0.0 is just a tiny rounding quirk) and every std is about 1. USArrests and Chickwts show 1.01 rather than 1.0 because pandas divides by n-1 while scikit-learn divides by n, and with only 50 and 71 rows the gap shows up in the second decimal. So there’s nothing wrong with your results.

**1f: Summarize the datasets**

```python
print("WINE")
print(wine["target"].value_counts())
display(wine.describe().T.round(2))

print("\nCHICKWTS")
display(chick.groupby("feed")["weight"].agg(["count", "mean", "std", "min", "max"]).round(1))

print("\nUSARRESTS")
display(arrests.describe().T.round(2))
```

Chickwts: sunflower has the highest average weight (328.9 g) and casein is close behind (323.6 g). Horsebean is the lowest (160.2 g). This matters for Step 3, because feeds with similar weight profiles, like sunflower and casein, should come out as similar in the recommender.
USArrests: Assault has by far the biggest raw numbers (mean 170.8, max 337), which is why we standardized it.

Sunflower and casein feeds gave the heaviest chicks; horsebean gave the lightest

**2a: Encode the labels and split the data**

```python
# 1. Convert class names to numbers
le = LabelEncoder()
y = le.fit_transform(wine["target"])
print(dict(zip(le.classes_, le.transform(le.classes_))))

# 2. Split: 80% train, 20% test, keeping class proportions equal
X_train, X_test, y_train, y_test = train_test_split(
    wine_features, y,
    test_size=0.2,
    stratify=y,
    random_state=RANDOM_STATE
)

print("Train:", X_train.shape, " Test:", X_test.shape)
print("Class counts in train:", np.bincount(y_train))
print("Class counts in test: ", np.bincount(y_test))
```

Why split before scaling? If we scaled the whole dataset first, information from the test bottles would leak into the training step, which is like seeing the exam questions before the exam. So from now on, the scaler and PCA learn only from the training set.

What each part does:

LabelEncoder turns class_0, class_1 and class_2 into 0, 1 and 2, which the model needs.
train_test_split sets aside 20% of the bottles as a test set the model never learns from.
stratify=y keeps the class mix the same in both sets, which matters because the classes are slightly uneven (59, 71, 48).

You should see roughly:

a mapping like {'class_0': 0, 'class_1': 1, 'class_2': 2}
shapes of (142, 13) for train and (36, 13) for test
class counts that are close to proportional in each set

**2b: Scale and apply PCA**

```python
# 1. Scale: learn from training data only
scaler = StandardScaler().fit(X_train)
X_train_s = scaler.transform(X_train)
X_test_s = scaler.transform(X_test)

# 2. PCA: keep enough components for 95% of the variance
pca = PCA(n_components=0.95, random_state=RANDOM_STATE).fit(X_train_s)
X_train_p = pca.transform(X_train_s)
X_test_p = pca.transform(X_test_s)

print("Original features:", X_train_s.shape[1])
print("Components kept for 95% variance:", pca.n_components_)
print("Variance per component:", pca.explained_variance_ratio_.round(3))
print("Total variance kept:", pca.explained_variance_ratio_.sum().round(3))
```

Scaling: we do this again, but now the scaler learns its averages and spreads from the training bottles only, then applies the same rule to the test bottles. That prevents data leakage.

PCA: the 13 measurements overlap a lot (for example, flavanoids and total phenols rise together). PCA squeezes them into fewer new columns, called components, that keep most of the information. It’s like summarizing 13 similar exam subjects as a few “overall strengths”. The rubric says to keep enough components to retain 95% of the variance (the information in the data).

NB:fit is only called on the training data, and the test data only goes through transform

**2c: Scree plot**

A scree plot shows how much variance each component carries, from biggest to smallest. It’s like a bar chart of how much each “overall strength” contributes. The left chart shows each component on its own, and the right chart shows the running total, with a red line at 95%.

```python
# Fit a PCA with ALL 13 components (training data only) just to draw the plot
full_pca = PCA().fit(X_train_s)
ratios = full_pca.explained_variance_ratio_
components = range(1, len(ratios) + 1)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Left: scree plot
ax[0].plot(components, ratios, "o-")
ax[0].set(title="Scree plot", xlabel="Principal component", ylabel="Explained variance ratio")

# Right: cumulative variance with the 95% target
ax[1].plot(components, np.cumsum(ratios), "o-")
ax[1].axhline(0.95, color="red", linestyle="--", label="95% target")
ax[1].set(title="Cumulative explained variance", xlabel="Number of components", ylabel="Cumulative variance")
ax[1].legend()

plt.tight_layout()
plt.show()
```

The first 2 components explain about 55% of the variance. 10 components are needed to reach 95%.

**2d: GridSearchCV to find the best k and distance metric**

k-NN has two settings you have to choose:

k: how many neighbours get a vote. With k=1 a single odd bottle can mislead the model, and with k=50 the vote gets blurry.
metric: how “distance” between bottles is measured.

GridSearchCV tries every combination for you.  It rates each combination with 5-fold cross-validation: the training data is cut into 5 parts, the model trains on 4 and is scored on the fifth, and this repeats so every part gets a turn. The test set is never touched.

```python
param_grid = {
    "n_neighbors": list(range(1, 26)),
    "metric": ["euclidean", "manhattan", "chebyshev"]
}

grid = GridSearchCV(
    KNeighborsClassifier(),
    param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
grid.fit(X_train_p, y_train)

print("Best parameters:", grid.best_params_)
print("Best cross-validation accuracy:", round(grid.best_score_, 4))

# Top 5 combinations
results = pd.DataFrame(grid.cv_results_)
display(results.sort_values("rank_test_score")[["param_n_neighbors", "param_metric", "mean_test_score"]].head(5))
```

What it does:

It tests 25 values of k with 3 distance metrics, which is 75 combinations.
Each is scored 5 times, so that’s 375 model fits. It should finish in a few seconds.
It uses X_train_p, the scaled and PCA-reduced training data from Step 2B, and nothing from the test set.

The metrics measure distance differently:

euclidean: the straight-line distance, “as the crow flies”.
manhattan: the distance along a grid of streets, adding up the differences in each direction.
chebyshev: just the single biggest differenc

The best combination is euclidean distance with k = 18, scoring 97.9% in cross-validation. In the table, the first column (19, 23, 17...) is just a row number. The real k values are in the second column: 20, 24, 18, 21 and 23.

Five combinations tie at exactly 0.979064: euclidean with k of 18, 20 and 24, and manhattan with k of 21 and 23. On a dataset this small, a one-bottle difference moves the score, so several settings tie. GridSearchCV breaks the tie by picking the first in its ranking, which is why I got k = 18.  The tie also shows the model isn’t sensitive to the exact k anywhere in the high teens or low twenties.

**2e: Train the final model and test it**

Now I'll use k = 18 with euclidean distance, fit it on the training bottles, and check it on the 36 test bottles it has never seen. This is the real exam, after the practice rounds of cross-validation.

```python
# Train the final model with the best parameters
best_knn = KNeighborsClassifier(**grid.best_params_)
best_knn.fit(X_train_p, y_train)

# Predict the unseen test bottles
y_pred = best_knn.predict(X_test_p)

print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print()
print(classification_report(y_test, y_pred, target_names=le.classes_))

# Confusion matrix
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt="d", cmap="Blues",
            xticklabels=le.classes_, yticklabels=le.classes_)
plt.title("k-NN confusion matrix (test set)")
plt.xlabel("Predicted class")
plt.ylabel("Actual class")
plt.show()
```

k-NN with PCA (10 components, 96.2% variance) and GridSearchCV (best: k=18, euclidean, CV accuracy 97.9%) achieved 100% accuracy on the 36-bottle test set. The three classes are well separated, but with a small test set the result may vary with a different split.

**3: Chickwts feed recommender**

The rubric wants this done by standardizing the weight, applying PCA to get one principal component, and computing cosine similarity between feed types.

 Cosine similarity compares the direction of two lists of numbers. If each feed is described by a single number, the answer can only be +1 or -1, which tells nothing. So i'll describe each feed with a profile of several numbers (average, median, spread, lowest and highest weight), using the standardized weight. Then I do both parts: PCA to one component, and cosine similarity on the full profile.

**3a: Build a profile for each feed**

```python
# One row per feed, describing its chicks' standardized weights
profile = chick.groupby("feed")["weight_scaled"].agg(["mean", "median", "std", "min", "max"])
print(profile.round(3))
```

Sunflower has the highest mean (0.872), with casein just behind at 0.803. Horsebean has the lowest (-1.304).
Positive means a feed’s chicks weighed more than average, and negative means less.

**3b: Compress each feed to one principal component**

Each feed is currently described by 5 numbers. PCA with n_components=1 squeezes them into a single “overall growth score” per feed, like turning five subject marks into one overall ranking. Because the 5 columns are on different scales, I standardize them first.

```python
# 1. Standardize the 5 profile columns so each counts equally
profile_scaled = StandardScaler().fit_transform(profile)

# 2. PCA down to ONE component
pca_feed = PCA(n_components=1, random_state=RANDOM_STATE)
feed_pc1 = pca_feed.fit_transform(profile_scaled)

print("Variance explained by PC1:", pca_feed.explained_variance_ratio_.round(3))

# 3. Show each feed's score, ranked
pc1_df = pd.DataFrame(feed_pc1, index=profile.index, columns=["PC1"]).sort_values("PC1")
display(pc1_df.round(3))

# 4. Quick bar chart
pc1_df["PC1"].plot(kind="barh", color="steelblue", title="Feed score on principal component 1")
plt.xlabel("PC1 score")
plt.show()
```

PC1 explains 86.8% of the differences between feeds, so one number summarizes them well. The ranking runs from horsebean (lowest) through linseed, soybean and meatmeal to sunflower and casein (highest).

Casein edges out sunflower on PC1 even though sunflower has the higher average weight. That’s because PC1 blends all 5 profile columns (average, median, spread, lowest, highest), not just the mean. The two feeds are very close either way.

**3c: Cosine similarity and recommendations**

Cosine similarity measures whether two feeds have the same shape of profile. It runs from -1 (opposite) through 0 (unrelated) to +1 (same pattern).

```python
# 1. Cosine similarity on PC1 only (the rubric version)
sim_pc1 = pd.DataFrame(cosine_similarity(feed_pc1),
                       index=profile.index, columns=profile.index)
print("Cosine similarity on PC1 alone:")
display(sim_pc1.round(2))

# 2. Cosine similarity on the full standardized profile (the useful version)
sim = pd.DataFrame(cosine_similarity(profile_scaled),
                   index=profile.index, columns=profile.index)

sns.heatmap(sim, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Cosine similarity between feed types")
plt.show()

# 3. Recommendation function: the feeds most similar to a chosen feed
def recommend(feed, n=3):
    return sim[feed].drop(feed).sort_values(ascending=False).head(n)

for f in profile.index:
    print(f"\nFeeds most similar to '{f}':")
    print(recommend(f).round(3).to_string())
```

Higher-growth group: casein and sunflower are the closest pair (0.751), and meatmeal sits loosely near casein (0.534).
Lower-growth group: horsebean, linseed and soybean are all tightly related. Linseed and soybean are almost identical (0.964), and horsebean and linseed are close behind (0.939).
Across the groups: the scores are negative, for example casein and soybean (-0.825) and sunflower and horsebean (-0.754). They are opposites in profile.

Meatmeal is the odd one out, with only one real match (casein, 0.534). It sits in the middle, which fits its average weight being near the overall mean.

One fix for the function: it always returns 3 feeds, so negative scores like -0.825 appear in the output as “recommendations”. That would look odd to a grader.

**3d: Cleaner recommendations**

```python
def recommend(feed, n=3):
    scores = sim[feed].drop(feed).sort_values(ascending=False)
    scores = scores[scores > 0].head(n)      # only genuinely similar feeds
    return scores

for f in profile.index:
    print(f"\nFeeds similar to '{f}':")
    result = recommend(f)
    print(result.round(3).to_string() if len(result) else "No similar feeds found")
```

Cosine similarity on PC1 alone can only return +1 or -1, so similarity was also computed on each feed's full standardized profile (mean, median, std, min, max of weight). Casein and sunflower (0.75) are the closest high-growth feeds, and linseed, soybean and horsebean form a tight low-growth group (0.85 to 0.96). Meatmeal is the least similar to the others, with casein as its only close match (0.53). A farmer using sunflower feed could consider casein as a substitute, while horsebean should be replaced with linseed or soybean.

**4: Clustering USArrests (K-Means and GMM)**

Clustering sorts the 50 states into groups with similar crime patterns, without being told what the groups are. The rubric wants these steps:

Standardize the data (already done in Step 1e as arrests_scaled).
Pick the top 3 most useful features.
Reduce them to 2 principal components then draw them on a flat chart.
Choose the number of clusters: the elbow method for K-Means, and BIC for GMM.
Fit both models and compare.

**4a: Pick the top 3 features and reduce to 2 components**

USArrests has 4 columns, so we keep the 3 that carry the most information. Two caveats: after standardizing, every column has the same spread, so we can’t just rank by variance. Instead we run PCA on all 4 and see which columns matter most to the main components, weighting each component by how much variance it explains. Then i'll run PCA again on the chosen 3, down to 2 components.

```python
# 1. Rank the 4 features by their importance in PCA
pca_all = PCA().fit(arrests_scaled)
importance = pd.Series(
    (np.abs(pca_all.components_.T) * pca_all.explained_variance_ratio_).sum(axis=1),
    index=arrests_scaled.columns
).sort_values(ascending=False)

print(importance.round(3))

# 2. Keep the top 3
top3 = importance.index[:3].tolist()
print("\nSelected features:", top3)

# 3. PCA down to 2 components
pca2 = PCA(n_components=2, random_state=RANDOM_STATE)
X2 = pca2.fit_transform(arrests_scaled[top3])

print("Variance per component:", pca2.explained_variance_ratio_.round(3))
print("Total variance kept:", pca2.explained_variance_ratio_.sum().round(3))
```

UrbanPop was dropped, since it scored lowest (0.428). The remaining three (Murder, Assault and Rape) are the three actual crime types, so the choice is sensible. The scores are close together (0.428 to 0.494), so this ranking is a narrow call rather than a big gap. You can mention that in your notes.

The 2 components keep 93.9% of the information. PC1 alone carries 78.6%, which is a “overall crime level” score, and PC2 adds 15.3%. That’s plenty to draw a faithful 2D picture of the states.

**4b: How many clusters?**

I'll have to tell the models how many groups to look for, so we test 1 to 10 and see which count fits best. K-Means and GMM use different tests.

Elbow method (K-Means): K-Means reports inertia, which is how spread out each cluster is. More clusters always lowers it, but the drop slows down. The “elbow” is where the line bends, and past it extra clusters add little. It’s like adding more waiters to a restaurant: the first few help a lot, and later ones barely change the wait.
BIC (GMM): the Bayesian Information Criterion scores how well the model fits while penalizing extra complexity. Lower is better, and the lowest point is the pick.

```python
ks = range(1, 11)

# K-Means: inertia for each number of clusters
inertia = [
    KMeans(n_clusters=k, n_init=10, random_state=RANDOM_STATE).fit(X2).inertia_
    for k in ks
]

# GMM: BIC for each number of components
bic = [
    GaussianMixture(n_components=k, n_init=5, random_state=RANDOM_STATE).fit(X2).bic(X2)
    for k in ks
]
best_k_gmm = int(np.argmin(bic)) + 1

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].plot(ks, inertia, "o-")
ax[0].set(title="K-Means elbow (inertia)", xlabel="Number of clusters (k)", ylabel="Inertia")

ax[1].plot(ks, bic, "o-", color="darkorange")
ax[1].axvline(best_k_gmm, color="red", linestyle="--", label=f"Lowest BIC at k={best_k_gmm}")
ax[1].set(title="GMM BIC (lower is better)", xlabel="Number of components", ylabel="BIC")
ax[1].legend()

plt.tight_layout()
plt.show()

print("Lowest BIC at k =", best_k_gmm)
```

Left chart (K-Means elbow): inertia falls steeply from k=1 (about 140) to k=2 (about 47), then drops only slowly: about 33 at k=3 and about 20 at k=4, flattening after that. The sharpest bend is at k = 2. There’s a weaker second bend around k=4.

Right chart (GMM BIC): the lowest point is clearly at k = 2 (about 293), and BIC rises after that.

Both methods agree on 2 clusters, which is a strong and easy-to-defend result for your write-up. Use best_k_kmeans = 2 so K-Means and GMM can be compared on the same number of groups.

**4c: Fit both models**

```python
# Fit both models. The elbow bends at k=2, which matches the BIC result.
best_k_kmeans = 2

# K-Means: hard assignments (each state belongs to exactly one cluster)
kmeans = KMeans(n_clusters=best_k_kmeans, n_init=10, random_state=RANDOM_STATE).fit(X2)

# GMM: soft assignments (each state gets a probability for every cluster)
gmm = GaussianMixture(n_components=best_k_gmm, n_init=5, random_state=RANDOM_STATE).fit(X2)

results = pd.DataFrame(X2, columns=["PC1", "PC2"], index=arrests.index)
results["kmeans"] = kmeans.labels_
results["gmm"] = gmm.predict(X2)
results["gmm_confidence"] = gmm.predict_proba(X2).max(axis=1)

print("K-Means cluster sizes:\n", results["kmeans"].value_counts().sort_index().to_string())
print("\nGMM cluster sizes:\n", results["gmm"].value_counts().sort_index().to_string())
display(results.head(10).round(3))
```

The two models agree closely. K-Means gives 30 and 20 states, and GMM gives 31 and 19, so they differ by at most one state. Cluster 0 in one lines up with cluster 0 in the other, which doesn’t always happen because the labels are arbitrary.

The first 10 rows show a pattern: Alabama, Alaska, Arizona, California, Colorado, Florida and Georgia are in cluster 1 with high PC1 scores, while Arkansas, Connecticut and Delaware are in cluster 0 with low ones. PC1 is the “overall crime level” axis, so cluster 1 is likely the higher-crime group (about 20 states) and cluster 0 the lower-crime group. I’ll confirm that with real numbers in the next step.

The gmm_confidence values are 0.97 to 1.00 for these states, so GMM is very sure about them. Any uncertainty will be on the borderline states.

**4d: Visualize and compare the clusters**

First the charts: K-Means, GMM, and a third panel colored by GMM confidence.

```python
fig, ax = plt.subplots(1, 3, figsize=(18, 5))

sns.scatterplot(data=results, x="PC1", y="PC2", hue="kmeans", palette="Set1", s=70, ax=ax[0])
ax[0].scatter(*kmeans.cluster_centers_.T, c="black", marker="X", s=200)
ax[0].set_title(f"K-Means (k={best_k_kmeans})")

sns.scatterplot(data=results, x="PC1", y="PC2", hue="gmm", palette="Set2", s=70, ax=ax[1])
ax[1].set_title(f"GMM (k={best_k_gmm})")

sc = ax[2].scatter(results["PC1"], results["PC2"], c=results["gmm_confidence"], cmap="viridis", s=70)
plt.colorbar(sc, ax=ax[2], label="Max cluster probability")
ax[2].set_title("GMM assignment confidence")

for a in ax:
    a.set_xlabel("PC1")
    a.set_ylabel("PC2")
plt.tight_layout()
plt.show()
```

Then the real crime numbers for each cluster, how the two models overlap, and the least certain states:

```python
labelled = arrests.join(results[["kmeans", "gmm"]])

print("K-Means cluster averages:")
display(labelled.groupby("kmeans").mean().round(1))

print("GMM cluster averages:")
display(labelled.groupby("gmm").mean().round(1))

print("K-Means vs GMM overlap:")
display(pd.crosstab(results["kmeans"], results["gmm"]))

print("Least certain GMM assignments:")
display(results.sort_values("gmm_confidence").head(5)[["kmeans", "gmm", "gmm_confidence"]].round(3))
```

What each part says:

The charts: two clear groups on the PC1 axis. The black X marks are the K-Means centers.
The averages: each cluster’s mean Murder, Assault, UrbanPop and Rape. This is what lets you name the clusters, for example “high-crime” and “low-crime”.
The overlap table: how many states both models put in the same group. Counts on the diagonal mean agreement.
The least certain states: the borderline states, and the only ones where K-Means and GMM could disagree.

Cluster 1 is the higher-crime group and cluster 0 the lower-crime group, as the PC1 scores suggested.


The gap is large. The higher-crime group has about 2.5 times the murder rate, over twice the assault rate and nearly twice the rape rate.

UrbanPop barely differs (63.6 vs 68.4). That supports dropping it in Step 4A: city size doesn’t separate the groups, while the three crime columns do.

The two models agree on 49 of 50 states. The overlap table puts 30 states in the low-crime group and 19 in the high-crime group under both models.

The one disagreement is Missouri. K-Means puts it in the high-crime group, while GMM puts it in the low-crime group, with only 88.4% confidence.
Other borderline states: Illinois, Texas, Tennessee and New York have the next lowest GMM confidence (0.89 to 0.96). They sit near the boundary between the groups, but GMM still assigns them to the high-crime group.

(A column to ignore: the gmm column inside the K-Means averages table is just the average of the GMM labels. It only confirms the near-perfect agreement.)

**4e: Write the interpretation**

The elbow method (K-Means) and BIC (GMM) both pointed to 2 clusters. After standardizing and keeping the top 3 features (Murder, Assault, Rape), PCA reduced the data to 2 components (93.9% of variance). K-Means found 30 and 20 states; GMM found 31 and 19, and the two models agree on 49 of 50 states. The higher-crime cluster averages 12.2 murders, 255 assaults and 29.2 rapes per 100,000 people, versus 4.9, 114 and 15.9 in the lower-crime cluster. Urban population barely differs between clusters (68.4 vs 63.6). Missouri is the only state the models disagree on, and GMM's confidence for it (88%) shows it is a borderline case. For stakeholders, GMM's probabilities are more useful than K-Means' hard labels for deciding which states need closer review.
