# Loan Approval Model: FinTech Innovations


## 1. Business Understanding


```python
cost_fp = 50_000   # approve a loan that should have been denied (default)
cost_fn = 8_000    # deny a loan that should have been approved (lost profit)

# Break-even: approve only if P(approved) is above this
breakeven = cost_fp / (cost_fp + cost_fn)
print(f"Approve only when P(approved) > {breakeven:.3f}")
print(f"One false approval costs as much as {cost_fp / cost_fn:.2f} false denials")
```

```python
import matplotlib.pyplot as plt
plt.bar(['False denial\n(lost profit)', 'False approval\n(default loss)'],
        [cost_fn, cost_fp], color=['#4C9F70', '#C0504D'])
plt.ylabel('Cost per mistake ($)')
plt.title('Cost of each error type')
plt.show()
```

```python
df.LoanApproved.value_counts(normalize=True)
from sklearn.metrics import roc_auc_score
roc_auc_score(df.LoanApproved, df.RiskScore)
```

### Business problem
FinTech Innovations wants a model that predicts whether a loan application should be approved, so the process is faster and more consistent than manual review by loan officers. The model also has to be explainable because of regulations.

### Cost of mistakes
I am treating LoanApproved = 1 as the positive class. There are two ways the model can be wrong.

A false positive means the model approves a loan that should have been denied. The company loses about $50,000 on a default.

A false negative means the model denies a loan that should have been approved. The company loses about $8,000 in profit.

So one bad approval costs 6.25 times as much as one wrongful denial. If I set the expected costs equal, the model should only approve when the probability of approval is above 50,000 / 58,000, which is about 0.86. The usual 0.5 cutoff would approve too many risky loans, so I will tune the threshold using cost instead.

### Metrics
About 76% of the applications were denied and 24% approved. Because of this, accuracy can be misleading. A model that denies everyone would still be around 76% accurate. I will use these instead

- Precision for approved loans, because it shows how many approvals were actually good ones and protects against the $50,000 mistake
- Recall, because it shows how many good applicants we still approve
- ROC-AUC, to see how well the model separates the two groups
- Total dollar cost (false positives x $ 50,000 + false negatives x $ 8,000), compared against approving everyone and denying everyone

## 5. Evaluation

```python
#Statistical tests

# Do approved and denied applicants differ on key numbers? (Mann-Whitney, no normality assumption)
for c in ['AnnualIncome', 'CreditScore', 'TotalDebtToIncomeRatio', 'LoanAmount']:
    a = df.loc[df.LoanApproved == 1, c]
    b = df.loc[df.LoanApproved == 0, c]
    print(c, "p-value:", stats.mannwhitneyu(a, b).pvalue)

# Is bankruptcy history related to approval? (chi-square)
table = pd.crosstab(df['BankruptcyHistory'], df['LoanApproved'])
print(table)
print("chi-square p-value:", stats.chi2_contingency(table)[1])
```

## 6. Conclusion and Executive Summary

###  Feature engineering ideas

These are ideas I want to try later in the pipeline. I have not built them yet.

1. Loan amount divided by annual income. This shows how big the loan is compared to what the person earns. A big loan for a small income should look riskier than the loan amount alone.

2. Monthly payment divided by monthly income. I plan to drop MonthlyLoanPayment because it may be set after the lender has already judged the risk, so I will estimate the payment as LoanAmount divided by LoanDuration instead, then divide that by MonthlyIncome.

3. Savings plus checking balance, compared to the loan amount. This tells me how much cash a person has to fall back on.

4. Liabilities divided by assets. NetWorth already looks like assets minus liabilities in most rows I checked, so I would not add that one again, but the ratio gives something different.

5. Log transform for LoanAmount, TotalAssets and TotalLiabilities. The boxplots showed some very large values, so a log transform could stop a few extreme rows from dominating.

6. EducationLevel should be ordered (High School, Associate, Bachelor, Master, Doctorate) instead of one-hot encoded, because there is a natural order.

NetWorth also does not equal TotalAssets minus TotalLiabilities in every row, so I will check how many rows differ before I trust it.
